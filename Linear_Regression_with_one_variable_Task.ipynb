Importing libraries necessary for the Task

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([ 0.        ,  0.52631579,  1.05263158,  1.57894737,  2.10526316,
     2.63157895,  3.15789474,  3.68421053,  4.21052632,  4.73684211,
     5.26315789,  5.78947368,  6.31578947,  6.84210526,  7.36842105,
     7.89473684,  8.42105263,  8.94736842,  9.47368421, 10.        ])

Y = np.array([ 2.60943416,  1.76213705,  7.18511292,  9.40744522,  5.46635068,
     8.6061673 , 13.30831239, 14.26225166, 16.70323979, 16.87285951,
    22.17984858, 23.81874177, 24.23732455, 28.20185083, 28.72449237,
    27.91299402, 32.21118578, 31.39802427, 36.91479534, 36.90014818])
```

# First step is to calculate the predicted **values**

```python
def y_pred(X, m ,c):
    return m * X + c
```

# Second Step: Function for Computing the error

```python
def compute_mse(y_true, y_pred):
  """Mean Squared Error"""
  return np.mean((y_true - y_pred)**2)
```

## Function to compute the gradient

```python
def gradients(x, y, m, c):
  n = x.shape[0]
  y_hat = m * x + c
  error = y - y_hat
  dm = (-2.0 / n) * np.sum(x * error)
  dc = (-2.0 / n) * np.sum(error)
  return dm, dc
```

# **Declaring Parameters**

```python
LR = 0.01   # Learning rate
m = 0
c = 0
epochs = 500

ms = []
cs = []
errors = []
```

# Training the model using gradient descent

```python
for i in range(epochs):
  Y_hat = y_pred(X, m, c)
  error = compute_mse(Y, Y_hat)
  mse = errors.append(error)
  grad = gradients(X, Y, m, c)

  ms.append(m)
  cs.append(c)

  print(f"Epoch {i+1}: MSE = {error}")

 # Parameter adjustment
  m = m - LR * grad[0]
  c = c - LR * grad[1]





```

# Cost Function Graph as Training progress

```python
plt.plot(X, Y, 'o')
for mi, ci in zip(ms, cs):
  plt.plot(X, mi * X + ci, '--')
plt.show()
```

# Final Value for model parameters m and c

```python
print("The final value for m =", m)
print("The final value for c =", c)
```

#Scatter plot of the original X and Y. Along with final linear model after training

```python
plt.figure(figsize=(8, 6))
plt.plot(X, Y, 'bo')
plt.plot(X, y_pred(X, m ,c), 'r--')
plt.show()
```
