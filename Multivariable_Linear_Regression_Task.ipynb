```python
df = pd.read_csv("/kaggle/input/multiple-linear-regression-dataset/multiple_linear_regression_dataset.csv")
df.head()
```

```python
print(df.shape)
df.isnull().sum()  # Check for missing values

df = df.dropna()  # Remove rows where any column is NaN
```

```python
import numpy as np

# Assuming the CSV has columns named 'x1', 'x2', 'y'
X1 = df['age'].values
X2 = df['experience'].values
Y = df['income'].values
```

A FUNCTION FOR THE PREDICTION

```python
def y_pred(X1, X2,  m1, m2, c):
  return m1 * X1 + m2 * X2 + c
```

FUNCTION TO COMPUTE COST FUNCTION

```python
def mse(Y, y_pred):
  return np.mean((Y - y_pred)**2)
```

INITIALIZING THE PARAMETERS

```python
m1, m2, c = 0, 0, 0
LR = 0.0001  # learning rate
epochs = 50
costs = []
```

COMPUTING THE GRADIENT AND UPDATING PARAMETERS

```python
for i in range(epochs):
  Y_pred = y_pred(X1, X2, m1, m2, c)
  cost = mse(Y, Y_pred)
  costs.append(cost)
  print(f"epoch {i+1}: MSE = {cost}")     #Print out the MSE after each cycle of training.


  # computing the gradients
  dm1 = -2 * np.mean(X1 * (Y - Y_pred))
  dm2 = -2 * np.mean(X2 * (Y - Y_pred))
  dc = -2 * np.mean(Y - Y_pred)

  # updating the parameters
  m1 = m1 - LR * dm1
  m2 = m2 - LR * dm2
  c = c - LR * dc

```

# Plot a graph of the cost function as training progresses

```python
  plt.plot(range(epochs), costs, 'ro')
  plt.title("Plot of cost function as training progresses")
  plt.show()
```

PRINTING THE FINAL VALUE FOR THE TRAINING

```python
print("m1", m1)
print("m2", m2)
print("c", c)
```
