# Analyzing each column and identifying data cleaning issue and propose appropriate fixes

```python
# Rank column
df['Rank'].isnull().value_counts()    # column is ready for training
```

```python
# Peak Column
df["Peak"].isnull().value_counts()
df.drop(columns=["Peak"], inplace=True) # drop the entire column

```

```python
# All time column
df["All Time Peak"].isnull().value_counts()
df["All Time Peak"].duplicated().value_counts()
df.drop(columns=["All Time Peak"], inplace=True)
```

```python
# what hidden characters
for col in df.columns:
    print(repr(col))
```

```python
# Actual gross column
df["Actual\xa0gross"] = df["Actual\xa0gross"].str.replace("$", "", regex=False) # strip the dolar sign
df["Actual\xa0gross"] = df["Actual\xa0gross"].str.replace(",", "", regex=False) # strip the comma
df["Actual\xa0gross"] = pd.to_numeric(df["Actual\xa0gross"], errors="coerce")
df["Actual\xa0gross"] = df["Actual\xa0gross"].fillna(df["Actual\xa0gross"].mean())  # fill the NaN wit
```

```python
# Adjusted gross (in 2022 dollars)
df["Adjusted\xa0gross (in 2022 dollars)"] = df["Adjusted\xa0gross (in 2022 dollars)"].str.replace("$", "", regex=False) # strip the dolar sign
df["Adjusted\xa0gross (in 2022 dollars)"] = df["Adjusted\xa0gross (in 2022 dollars)"].str.replace(",", "", regex=False) # strip the comma
df["Adjusted\xa0gross (in 2022 dollars)"] = pd.to_numeric(df["Adjusted\xa0gross (in 2022 dollars)"], errors="coerce")
```

```python
# Tour title column
df["Tour title"].isna()   # All the column is NaN
df["Tour title"].duplicated().value_counts()  # No duplicated
df.drop(columns=["Tour title"], inplace=True) # ordinal type

```

```python
# Artist column
df["Artist"].isna().value_counts()  # No NaN
df["Artist"].duplicated().value_counts()  # 11 duplicated
df.drop_duplicates(subset = ["Artist"], keep = 'first', inplace=True)

```

```python
# shows coumn
df["Shows"].isna().value_counts()  # No NaN
df["Shows"].duplicated().value_counts()  # one duplicated
df.drop_duplicates(inplace = True)
```

```python
# Year(s) column
df["Year(s)"] = df["Year(s)"].str.replace("–", "-", regex=False)
df["Start_Year"] = df["Year(s)"].str.split("-").str[0].astype(int)

df["End_Year"] = df["Year(s)"].str.split("-").str[1]
df["End_Year"] = df["End_Year"].fillna(df["Start_Year"]).astype(int)
df.drop(columns=["Year(s)"], inplace=True)
```

```python
# Average gross column
df["Average gross"] = df["Average gross"].str.replace("$", "", regex=False) # strip the dolar sign
df["Average gross"] = df["Average gross"].str.replace(",", "", regex=False) # strip the comma
df["Average gross"] = pd.to_numeric(df["Average gross"], errors="coerce")
```

```python
# Ref. column
df["Ref."].isna().value_counts()  # all the column is NaN
del df["Ref."]  # removing the entire column since it doesn't provide signinificant info. for the training
```

# Using sklearn to implement Feature Engineering

## - For categorical features, One hot encoding is used

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

```

```python
# Artist
categorical_features = ["Artist"]

# Rank, Actual\xa0gross, Adjusted\xa0gross (in 2022 dollars), Shows, Year(s), Average gross, column
numerical_features = ["Rank", "Actual\xa0gross", "Adjusted\xa0gross (in 2022 dollars)", "Shows", "Start_Year", "End_Year", "Average gross"]

numeric_transformer = StandardScaler()
categorical_transformer = OneHotEncoder(handle_unknown="ignore")

scaling_encoding = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

pipe = Pipeline(steps=[
    ('preprocessor', scaling_encoding)
])

# Fit pipeline on dataset
pipe.fit(df)
```
