# Initializing Training parameters and hyperparameters

```python
m1, m2, m3, c = 0.5, 0.5, 0.5, 0.0
alpha = 0.01
epochs = 1_000
costs = []
```

# Logistic (Sigmoid) Function

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

# Extracting X and Y

```python
Y = df['label'].values

df["x1"] = np.array(pd.to_numeric(df["x1"], errors="coerce"))
df["x2"] = np.array(pd.to_numeric(df["x2"], errors="coerce"))
df["x3"] = np.array(pd.to_numeric(df["x3"], errors="coerce"))

X = df[["x1", "x2", "x3"]].values
```

# cost for each data point and gradient for each parameter as well as updating the parameters

```python
for i in range(epochs):
    Y_pred = sigmoid(m1*df['x1'] + m2*df['x2'] + m3*df['x3'] + c)
    cost = -np.mean(Y * np.log(Y_pred) + (1 - Y) * np.log(1 - Y_pred))
    print(f"Epoch {i+1}: Loss = {cost:.5f}")    # Print out the binary cross-entropy loss after each cycle of training.
    costs.append(cost)

    # Gradient
    raw_residual = Y_pred - Y
    dmc = np.mean(raw_residual)
    dm1 = np.mean(raw_residual * df['x1'])
    dm2 = np.mean(raw_residual * df['x2'])
    dm3 = np.mean(raw_residual * df['x3'])

    # Update parameters
    c  = c - alpha * dmc
    m1 = m1 - alpha * dm1
    m2 = m2 - alpha * dm2
    m3 = m3 - alpha * dm3
```

# Graph of the cost function as training progresses

```python
plt.figure(figsize = (4, 3))
plt.plot(range(len(costs)), costs, 'o')
plt.show()
```
